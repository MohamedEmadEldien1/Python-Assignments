### Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.

```python
def check_season(month):
    month = month.capitalize()     

    if month in ['September', 'October', 'November']:
        return 'Autumn'
    elif month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    else:
        return 'Invalid month'

print(check_season('October'))    
print(check_season('january'))   
print(check_season('Mayy'))   
```

### Write a function named print_list. It takes a list as a parameter and it prints out each element of the list.

```python
def print_list(lst):
    for i in range(len(lst)):
        print(lst[i])
   
print_list(['Ali', 'Ahmed', 'Mona'])
```

### Write a function named reverse_list. It takes a list as a parameter and it returns the reverse of the list.

```python
def print_revlist(lst):
    for i in range(len(lst)):
        print(lst[-i-1])

print_revlist(['Ali', 'Ahmed', 'Mona'])
```

### Write a function named sum_of_numbers. It takes two parameters min_num and max_num and it adds all the numbers in that range.

```python
def sum_of_numbers(min_num, max_num):
    total = 0
    for n in range(min_num, max_num + 1):
        total += n
    return total

print(sum_of_numbers(1, 5))   
print(sum_of_numbers(3, 6))
```

### Write a function named sum_of_numbers. It takes two parameters min_num and max_num and it adds all the `odd` numbers in that range.

```python
def sum_of_odnumbers(min_num, max_num):
    total = 0
    for n in range(min_num, max_num + 1):
        if n % 2 != 0:
            total += n
    return total
```

### Write a function named sum_of_numbers. It takes two parameters min_num and max_num and it adds all the `even` numbers in that range.

```python
def sum_of_evnumbers(min_num, max_num):
    total = 0
    for n in range(min_num, max_num + 1):
        if n % 2 == 0:
            total += n
    return total
```

### Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.

    Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    Expected Output :
    No. of Upper case characters : 4
    No. of Lower case Characters : 33

```python
def calc_num_upandlow(s):
    up=0
    low=0
    for i in range (len(s)):
        if s[i].isupper():
            up +=1
        elif s[i].islower():
            low+=1
    print ("No. of Upper case characters :", up)
    print("No. of Lower case Characters :", low)

calc_num_upandlow('Hello Mr. Rogers, how are you this fine Tuesday?')
```

### Write a Python function that checks whether a passed in string is palindrome or not.

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g., `madam` or `nurses run`.

```python
def palindrome(x):
    r = x.replace(' ', '').lower()
    for i in range(0, len(r)):
        if r[i] != r[-i - 1]:
            return False
    return True
```
