## EX8 : write a program that accept from user 2 numbers and make simple calculator on it ?

```python
print("Enter first number >")
num1 = float(input())

print("Enter second number >")
num2 = float(input())

print("Choose operation (+, -, *, /) >")
op = input()

if op == "+":
    result = num1 + num2
elif op == "-":
    result = num1 - num2
elif op == "*":
    result = num1 * num2
elif op == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: division by zero"
else:
    result = "Error: invalid operation"

print("Result:", result)
```

## Ask the user for three numbers and print the largest one (using if-elif-else).

```python


nu1 = int(input('write the 1st number '))
nu2 = int(input('write the 2nd number '))
nu3 = int(input('write the 3rd number '))

if nu1 > nu2 and nu1 > nu3:
    print("Largest Number is ", nu1)
elif nu2 > nu1 and nu2 > nu3:
    print("Largest Number is ", nu2)
elif nu3 > nu1 and nu3 > nu2:
    print("Largest Number is ", nu3)
```
