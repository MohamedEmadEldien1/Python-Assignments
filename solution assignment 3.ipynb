## EX4: ask user to enter number and show this number is even or not ?

## EX5: ask user to enter number and show whether this number is divisible by 2 and 3 ?

```python
z=int(input('input number '))
if (z % 2)==0 and (z % 3)==0:
    print('divisible by 2 and 3')
elif (z % 2)==0:
    print('not divisible by 3')
elif (z % 3)==0:
    print('not divisible by 2')
else:
    print ('not divisible by 2 and 3')
```

## EX6: ask user to enter two numbers and operation and perform this operation ?

```python
x=int(input('input 1st '))
y=int(input('input 2nd '))
op=input('inter the operation (* , + , - or / )')
if op == '+':
    result = x + y
elif op == '-':
    result = x - y
elif op == '*':
    result = x * y
elif op == '/':
    if y != 0:
        result = x / y
    else:
        result = 'Error: division by zero'
else:
    result = 'Error: invalid operation'

print('Result:', result)
```

## EX12: ask user to enter string and if that string contain only numbers convert it to integer ?

```python
text = input('Enter a string >')

if text.isdigit():
    number = int(text)
    print('Converted to integer:', number)
else:
    print('String contains non-numeric characters, cannot convert')
```

## EX14 : ask user to enter binary number (string contain (0 , 1)) then check whether this is valid binary number or not ?

```python
var = input('var')

if set(var).issubset({'0', '1'}):
    print('Valid binary number')
else:
    print('Not a valid binary number')
```
