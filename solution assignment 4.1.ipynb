## **Problem 1: Basic Arithmetic Operations**
- Write a program that takes two integers with variables names `a` and `b`, and prints their sum, difference, multiplication, exponent, division, floor division and modulus.

```python
# Answer here
a = int(input("Enter a: "))
b = int(input("Enter b: "))

print("Sum:", a + b)
print("Difference:", a - b)
print("Multiplication:", a * b)
print("Exponent:", a ** b)

if b != 0:
    print("Division:", a / b)
    print("Floor division:", a // b)
    print("Modulus:", a % b)
else:
    print("Division, floor division and modulus are not possible when b is 0")
```
