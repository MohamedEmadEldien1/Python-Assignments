## **Problem 18: IF Condition**
#### Write a program that takes a user's age as input and determines their life stage.
* Print "Child" if the age is less than 13,
* "Teen" if the age is between 13 and 19,
* "Young Adult" if the age is between 20 and 35,
* and "Adult" if the age is 36 or older.

```python
# Answer here
age = int(input("Enter your age: "))

if age < 13:
    print("Child")
elif age <= 19:
    print("Teen")
elif age <= 35:
    print("Young Adult")
else:
    print("Adult")
    
```

## **Problem 19: IF Condition**
- Write a Python program to Check if a Triangle is (Equilateral متساوي الأضلاع), (Isosceles متساوي الساقين), or (Scalene مختلف الأضلاع).
- Take `a`, `b`, `c` as inputs from the user

```python
# Answer here
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

if a == b == c:
    print("Equilateral")
elif a == b or b == c or a == c:
    print("Isosceles")
else:
    print("Scalene")
```
